## 🎯 Experiment 1: Train with Different Batch Sizes

Let's track experiments with varying batch sizes to find the optimal configuration.

```python
# ============================================================================
# Experiment 1: Training with Different Batch Sizes
# ============================================================================
# GOAL: Find the optimal batch size for training RipCatch

# WHAT IS BATCH SIZE?
# ------------------
# Batch size is how many images the model processes at once during training.
# 
# ANALOGY: Imagine grading homework
# - Batch size 8 = Grade 8 papers, update teaching method, repeat
# - Batch size 32 = Grade 32 papers, update teaching method, repeat
#
# TRADE-OFFS:
# - Smaller batch (8, 16): More updates, slower training, less GPU memory
# - Larger batch (32, 64): Faster training, more GPU memory needed
#
# WHY TEST DIFFERENT BATCH SIZES?
# - Find the sweet spot between speed and accuracy
# - Avoid running out of GPU memory
# - Optimize training efficiency

# List of batch sizes to try
batch_sizes = [8, 16, 32]  # Start with common values

print("="*60)
print("🧪 EXPERIMENT: Batch Size Comparison")
print("="*60)
print(f"Testing batch sizes: {batch_sizes}")
print(f"Each will train for 10 epochs (quick test)")
print(f"\n💡 NOTE: In real training, you'd use 100+ epochs")
print("="*60 + "\n")

# Loop through each batch size and train
for batch_size in batch_sizes:
    
    print(f"\n{'='*60}")
    print(f"📊 Testing Batch Size: {batch_size}")
    print(f"{'='*60}\n")
    
    # Create a descriptive run name
    # This helps identify the experiment later
    run_name = f"batch_size_{batch_size}"
    
    # Start MLflow run for this experiment
    with mlflow.start_run(run_name=run_name) as run:
        
        # STEP 1: Log the batch size parameter
        # This records what setting we're testing
        mlflow.log_param("batch_size", batch_size)
        mlflow.log_param("epochs", 10)  # Just 10 epochs for quick testing
        mlflow.log_param("image_size", 640)
        mlflow.log_param("experiment_type", "batch_size_comparison")
        
        # STEP 2: Initialize trainer
        # (In real code, you'd actually train here)
        print(f"   🏋️  Training with batch size {batch_size}...")
        print(f"   ⏱️  This would normally take several minutes...")
        
        # STEP 3: Simulate training results
        # (Replace this with actual training in your real experiments)
        # Simulated metrics - normally these come from model.train()
        import random
        simulated_map = 0.85 + random.uniform(-0.05, 0.05)
        simulated_loss = 0.15 + random.uniform(-0.03, 0.03)
        
        # STEP 4: Log metrics
        mlflow.log_metric("mAP50", simulated_map)
        mlflow.log_metric("train_loss", simulated_loss)
        mlflow.log_metric("gpu_memory_mb", batch_size * 100)  # Approximate
        
        # STEP 5: Add tags for organization
        mlflow.set_tags({
            "experiment_series": "batch_size_study",
            "purpose": "optimization",
            "date": "2026-01-28"
        })
        
        print(f"   ✅ Training complete!")
        print(f"   📊 mAP@50: {simulated_map:.4f}")
        print(f"   📉 Loss: {simulated_loss:.4f}")
        print(f"   💾 Run ID: {run.info.run_id}\n")

print("\n" + "="*60)
print("✅ BATCH SIZE EXPERIMENTS COMPLETE!")
print("="*60)
print("\n💡 NEXT STEPS:")
print("   1. Open MLflow UI: http://localhost:5000")
print("   2. Select all three runs (checkboxes)")
print("   3. Click 'Compare' button")
print("   4. See which batch size performed best!")
print("   5. Look at charts to visualize differences")
print("="*60)
```
